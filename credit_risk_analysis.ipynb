```python
X = df.drop("TARGET", axis=1)
y = df["TARGET"]
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print(y.value_counts())
```

```python
X = X.drop("SK_ID_CURR", axis=1)
```

```python
print(X.shape)
```

```python
numeric_features = X.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_features = X.select_dtypes(
    include=["object"]
).columns
```

```python
print("Numerical columns:", len(numeric_features))
print("Categorical columns:", len(categorical_features))
```

```python
print("Categorical columns:")
print(categorical_features.tolist())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_valid, y_train, y_valid = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training data:", X_train.shape)
print("Validation data:", X_valid.shape)

print("\nTraining target:")
print(y_train.value_counts(normalize=True))

print("\nValidation target:")
print(y_valid.value_counts(normalize=True))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
```

```python
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)
```

```python
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)

X_valid_processed = preprocessor.transform(X_valid)
```

```python
print("Original training shape:", X_train.shape)
print("Processed training shape:", X_train_processed.shape)

print("Original validation shape:", X_valid.shape)
print("Processed validation shape:", X_valid_processed.shape)
```

```python
print(X_train_processed.shape)
print(X_valid_processed.shape)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(
            max_iter=1000,
            class_weight="balanced"
        ))
    ]
)
```

```python
logistic_model.fit(X_train, y_train)
```

```python
y_pred = logistic_model.predict(X_valid)
```

```python
y_proba = logistic_model.predict_proba(X_valid)[:, 1]
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)
```

```python
accuracy = accuracy_score(y_valid, y_pred)
precision = precision_score(y_valid, y_pred)
recall = recall_score(y_valid, y_pred)
f1 = f1_score(y_valid, y_pred)
roc_auc = roc_auc_score(y_valid, y_proba)

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)
print("ROC-AUC  :", roc_auc)
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_valid, y_pred)

print(cm)
```

```python
disp = ConfusionMatrixDisplay(confusion_matrix=cm)

disp.plot()

plt.title("Logistic Regression - Confusion Matrix")

plt.show()
```

```python
from sklearn.metrics import RocCurveDisplay

RocCurveDisplay.from_predictions(
    y_valid,
    y_proba
)

plt.title("Logistic Regression - ROC Curve")

plt.show()
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
random_forest_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", RandomForestClassifier(
            n_estimators=200,
            random_state=42,
            class_weight="balanced",
            n_jobs=-1
        ))
    ]
)
```

```python
random_forest_model.fit(X_train, y_train)
```

```python
rf_pred = random_forest_model.predict(X_valid)
```

```python
rf_proba = random_forest_model.predict_proba(X_valid)[:, 1]
```

```python
rf_accuracy = accuracy_score(y_valid, rf_pred)
rf_precision = precision_score(y_valid, rf_pred)
rf_recall = recall_score(y_valid, rf_pred)
rf_f1 = f1_score(y_valid, rf_pred)
rf_roc_auc = roc_auc_score(y_valid, rf_proba)

print("Random Forest Results")
print("---------------------")
print("Accuracy :", rf_accuracy)
print("Precision:", rf_precision)
print("Recall   :", rf_recall)
print("F1 Score :", rf_f1)
print("ROC-AUC  :", rf_roc_auc)
```

```python
cm_rf = confusion_matrix(y_valid, rf_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_rf
)

disp.plot()

plt.title("Random Forest - Confusion Matrix")

plt.show()
```

```python
from xgboost import XGBClassifier
```

```python
from xgboost import XGBClassifier

xgb_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", XGBClassifier(
            n_estimators=200,
            learning_rate=0.05,
            max_depth=6,
            random_state=42,
            eval_metric="logloss",
            n_jobs=-1
        ))
    ]
)
```

```python
xgb_model.fit(X_train, y_train)
```

```python
xgb_pred = xgb_model.predict(X_valid)
```

```python
xgb_proba = xgb_model.predict_proba(X_valid)[:, 1]
```

```python
xgb_accuracy = accuracy_score(y_valid, xgb_pred)
xgb_precision = precision_score(y_valid, xgb_pred)
xgb_recall = recall_score(y_valid, xgb_pred)
xgb_f1 = f1_score(y_valid, xgb_pred)
xgb_roc_auc = roc_auc_score(y_valid, xgb_proba)

print("XGBoost Results")
print("----------------")
print("Accuracy :", xgb_accuracy)
print("Precision:", xgb_precision)
print("Recall   :", xgb_recall)
print("F1 Score :", xgb_f1)
print("ROC-AUC  :", xgb_roc_auc)
```

```python
cm_xgb = confusion_matrix(y_valid, xgb_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_xgb
)

disp.plot()
plt.title("XGBoost - Confusion Matrix")
plt.show()
```

```python
model_results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest",
        "XGBoost"
    ],
    "Accuracy": [
        accuracy,
        rf_accuracy,
        xgb_accuracy
    ],
    "Precision": [
        precision,
        rf_precision,
        xgb_precision
    ],
    "Recall": [
        recall,
        rf_recall,
        xgb_recall
    ],
    "F1 Score": [
        f1,
        rf_f1,
        xgb_f1
    ],
    "ROC-AUC": [
        roc_auc,
        rf_roc_auc,
        xgb_roc_auc
    ]
})

print(model_results)
```

```python
model_results = model_results.round(4)

model_results
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    model_results["Model"],
    model_results["ROC-AUC"]
)

plt.title("Model Comparison - ROC-AUC")
plt.xlabel("Model")
plt.ylabel("ROC-AUC")
plt.xticks(rotation=15)
plt.show()
```
